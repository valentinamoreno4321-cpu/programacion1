```python
import re
```

```python
# 1. excepcion personalizada
class correoinvalidoerror(Exception):
    """excepcion lanzada cuando el correo no cumple el formato institucional."""
```

```python
# 2. funcion de validacion
def validar_correo(correo: str) -> bool:
    # regex: nombre(minusculas) _ edad(numeros) _ apellido(minusculas) @ dominio exacto
    patron = r"^[a-z]+_[0-9]+_[a-z]+@universitaria\.edu\.co$"

    if not re.fullmatch(patron, correo):
        raise correoinvalidoerror(
            "formato incorrecto. debe ser: nombre_edad_apellido@universitaria.edu.co\n"
            "solo se permiten letras minusculas, numeros y guion bajo (_) en la parte local."
        )
    return True
```

```python
# 3. programa principal con control de intentos
max_intentos = 3
intentos = 0

print("sistema de validacion de correo institucional")
print(f"tiene {max_intentos} intentos para ingresar un correo valido.\n")

while intentos < max_intentos:
    try:
        correo = input("ingrese su correo institucional: ").strip()
        validar_correo(correo)
        print("correo valido. registro exitoso.")
        break
    except correoinvalidoerror as e:
        intentos += 1
        restantes = max_intentos - intentos
        if restantes > 0:
            print(f"{e}")
            print(f"intento {intentos}/{max_intentos}. le quedan {restantes} intento(s).\n")
        else:
            print(f"{e}")
            print("\nacceso bloqueado por intentos fallidos.")
    except Exception as e:
        print(f"error inesperado: {type(e).__name__} - {e}")
        intentos += 1
```
