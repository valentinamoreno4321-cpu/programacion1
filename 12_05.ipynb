```python
usuarios = {
    "VALENTINA": "1234",
    "DIEGO" : "ABCD"
}

try:
    usuario = input("Usuario: ").upper()
    contraseña = input("Contraseña: ")

    if usuario not in usuarios:
        raise keyError("No existe ese usuario")
    if usuarios[usuario] != contraseña:
        raise ValueError("Contraseña incorrecta")
    else:
      print("Bienvenido/a,", usuario)

except KeyError as e:
    print ("Error:", e)
except ValueError as e:
    print ("Error:", e)
```

El error ocurre porque input() guarda los valores como texto (str) y no como números, por lo que Python no puede realizar operaciones matemáticas como resta, división o multiplicación entre cadenas. Por ejemplo, num1 - num2 genera un error porque ambos son texto.

```python
import sys

print("=== CALCULADORA ===")

num1 = input("Numero 1: ")
num2 = input("Numero 2: ")
op = input("Operación (+,-,*,/,%): ")

# Validar números
if not num1.replace(".", "", 1).isdigit():
    print("Error: El primer valor no es un número")
    sys.exit()

if not num2.replace(".", "", 1).isdigit():
    print("Error: El segundo valor no es un número")
    sys.exit()

num1 = float(num1)
num2 = float(num2)

# Validar operación
if op not in ["+", "-", "*", "/", "%"]:
    print("Error: Operación no válida")
    sys.exit()

# Validar división entre cero
if (op == "/" or op == "%") and num2 == 0:
    print("Error: No se puede dividir entre cero")
    sys.exit()

# Operaciones
if op == "+":
    resultado = num1 + num2

if op == "-":
    resultado = num1 - num2

if op == "*":
    resultado = num1 * num2

if op == "/":
    resultado = num1 / num2

if op == "%":
    resultado = num1 % num2

print("Resultado:", resultado)
```

```python
import re
texto = "Tengo 25 gatos y 1 perro"
resultado = re.findall(r"\d", texto)
print(resultado)
```

```python
import re
texto = "Tengo 25 gatos y 1 perro"
resultado = re.findall(r"\d+", texto)
print(resultado)
```
